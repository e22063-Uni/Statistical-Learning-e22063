# Kalman Filter Derivations

## Assumptions & Notation
We assume the process noise $w_{k-1}$ and measurement noise $z_k$ are zero-mean, independent Gaussian white noise processes:
$$w_{k-1} \sim \mathcal{N}(0, \Sigma_p)$$
$$z_k \sim \mathcal{N}(0, \Sigma_m)$$
where $w_{k-1}$ and $z_k$ are independent of each other and independent of the prior state $x^{+}_{k-1}$.

---

## 1. Derivation of the Predicted State Distribution $x_k^-$

We are given the state prediction equation:
$$x^{-}_k = A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}$$

Since $x^{+}_{k-1}$ and $w_{k-1}$ are normally distributed and the operation is a linear combination, the predicted state $x^{-}_k$ is also normally distributed:
$$x_k^- \sim \mathcal{N}(m_k^-, P_k^-)$$

### **Predicted Mean ($m_k^-$):**
Taking the mathematical expectation on both sides:
$$
\begin{aligned}
m_k^- = \mathbb{E}[x^{-}_k] &= \mathbb{E}[A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}] \\
&= A_{k-1}\mathbb{E}[x^{+}_{k-1}] + G_{k-1}\mathbb{E}[w_{k-1}]
\end{aligned}
$$
Since $\mathbb{E}[x^{+}_{k-1}] = m_{k-1}$ and $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = A_{k-1}m_{k-1}$$

### **Predicted Covariance ($P_k^-$):**
The error framework for $x^{-}_k$ around its mean is:
$$x^{-}_k - m_k^- = A_{k-1}(x^{+}_{k-1} - m_{k-1}) + G_{k-1}w_{k-1}$$

The covariance matrix is defined as $P_k^- = \mathbb{E}[(x^{-}_k - m_k^-)(x^{-}_k - m_k^-)^T]$:
$$
\begin{aligned}
P_k^- &= \mathbb{E}\left[\left(A_{k-1}(x^{+}_{k-1} - m_{k-1}) + G_{k-1}w_{k-1}\right)\left(A_{k-1}(x^{+}_{k-1} - m_{k-1}) + G_{k-1}w_{k-1}\right)^T\right] \\
&= A_{k-1}\mathbb{E}[(x^{+}_{k-1} - m_{k-1})(x^{+}_{k-1} - m_{k-1})^T]A_{k-1}^T + G_{k-1}\mathbb{E}[w_{k-1}w_{k-1}^T]G_{k-1}^T
\end{aligned}
$$
*(Note: The cross-terms vanish because $x^{+}_{k-1}$ and $w_{k-1}$ are independent).*

Substituting the known covariances $\text{Var}(x^{+}_{k-1}) = P_{k-1}$ and $\text{Var}(w_{k-1}) = \Sigma_p$:
$$P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T$$

---

## 2. Derivation of the Predicted Measurement Distribution $y_k^-$

The measurement model is given by:
$$y^{-}_k = H_k\,x^{-}_k + z_k$$

Because $x^{-}_k$ and $z_k$ are Gaussian and the transformation is linear, $y^{-}_k$ is also Gaussian:
$$y_k^- \sim \mathcal{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m)$$

### **Expected Measurement Mean:**
$$
\begin{aligned}
\mathbb{E}[y^{-}_k] &= \mathbb{E}[H_k\,x^{-}_k + z_k] \\
&= H_k \mathbb{E}[x^{-}_k] + \mathbb{E}[z_k]
\end{aligned}
$$
Since $\mathbb{E}[x^{-}_k] = m_k^-$ and $\mathbb{E}[z_k] = 0$:
$$\mathbb{E}[y^{-}_k] = H_k m_k^-$$

### **Predicted Measurement Covariance:**
The measurement error relative to its mean is:
$$y^{-}_k - H_k m_k^- = H_k(x^{-}_k - m_k^-) + z_k$$

The covariance matrix is defined as $\text{Var}(y^{-}_k) = \mathbb{E}[(y^{-}_k - H_k m_k^-)(y^{-}_k - H_k m_k^-)^T]$:
$$
\begin{aligned}
\text{Var}(y^{-}_k) &= \mathbb{E}\left[\left(H_k(x^{-}_k - m_k^-) + z_k\right)\left(H_k(x^{-}_k - m_k^-) + z_k\right)^T\right] \\
&= H_k \mathbb{E}[(x^{-}_k - m_k^-)(x^{-}_k - m_k^-)^T]H_k^T + \mathbb{E}[z_k z_k^T]
\end{aligned}
$$
*(Note: Cross-terms vanish because the measurement noise $z_k$ is independent of the predicted state $x^{-}_k$).*

Substituting $\text{Var}(x^{-}_k) = P_k^-$ and $\text{Var}(z_k) = \Sigma_m$:
$$\text{Var}(y^{-}_k) = H_k P_k^- H_k^T + \Sigma_m$$

Thus:
$$y_k^- \sim \mathcal{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m)$$

---

## 3. Derivation of the Joint Distribution of $x_k^-$ and $y_k^-$

Since $y_k^-$ depends linearly on $x_k^-$, they are jointly Gaussian:
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathcal{N}\left( \mu_{\text{joint}}, \Sigma_{\text{joint}} \right)$$

### **Joint Mean Vector:**
$$\mu_{\text{joint}} = \begin{bmatrix} \mathbb{E}[x_k^-] \\ \mathbb{E}[y^{-}_k] \end{bmatrix} = \begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix}$$

### **Joint Covariance Matrix:**
$$\Sigma_{\text{joint}} = \begin{bmatrix} \text{Var}(x_k^-) & \text{Cov}(x_k^-, y_k^-) \\ \text{Cov}(y_k^-, x_k^-) & \text{Var}(y_k^-) \end{bmatrix}$$

We already know $\text{Var}(x_k^-) = P_k^-$ and $\text{Var}(y_k^-) = H_k P_k^- H_k^T + \Sigma_m$. Let's compute the cross-covariance $\text{Cov}(x_k^-, y_k^-)$:
$$
\begin{aligned}
\text{Cov}(x_k^-, y_k^-) &= \mathbb{E}[(x_k^- - m_k^-)(y_k^- - H_k m_k^-)^T] \\
&= \mathbb{E}[(x_k^- - m_k^-) (H_k(x_k^- - m_k^-) + z_k)^T] \\
&= \mathbb{E}[(x_k^- - m_k^-)(x_k^- - m_k^-)^T]H_k^T + \mathbb{E}[(x_k^- - m_k^-)z_k^T]
\end{aligned}
$$
Because $x_k^-$ and $z_k$ are independent, $\mathbb{E}[(x_k^- - m_k^-)z_k^T] = 0$. This leaves:
$$\text{Cov}(x_k^-, y_k^-) = P_k^- H_k^T$$

Since $\text{Cov}(y_k^-, x_k^-) = \left[\text{Cov}(x_k^-, y_k^-)\right]^T$:
$$\text{Cov}(y_k^-, x_k^-) = H_k P_k^-$$

Assembling the complete joint matrix yields:
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathcal{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right)$$

---

## 4. Derivation of the Conditional State Distribution (Kalman Update)

To track the conditional distribution $x^{+}_k \triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$, we use the standard rule for **Conditional Multivariate Gaussian Distributions**.

If a joint distribution is given by:
$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} \mu_X \\ \mu_Y \end{bmatrix}, \begin{bmatrix} \Sigma_{XX} & \Sigma_{XY} \\ \Sigma_{YX} & \Sigma_{YY} \end{bmatrix} \right)$$

Then the conditional distribution of $X$ given $Y = y$ is $\mathcal{N}(\mu_{X|Y}, \Sigma_{X|Y})$, where:
$$\mu_{X|Y} = \mu_X + \Sigma_{XY} \Sigma_{YY}^{-1} (y - \mu_Y)$$
$$\Sigma_{X|Y} = \Sigma_{XX} - \Sigma_{XY} \Sigma_{YY}^{-1} \Sigma_{YX}$$

### **Mapping to our variables:**
* $X = x_k^- \implies \mu_X = m_k^-$ and $\Sigma_{XX} = P_k^-$
* $Y = y_k^- \implies \mu_Y = H_k m_k^-$ and $\Sigma_{YY} = H_k P_k^- H_k^T + \Sigma_m$
* $\Sigma_{XY} = P_k^- H_k^T$ and $\Sigma_{YX} = H_k P_k^-$

### **Defining the Kalman Gain ($K_k$):**
Let $K_k = \Sigma_{XY} \Sigma_{YY}^{-1}$:
$$K_k = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$

### **Updated Mean ($m_k$):**
$$m_k = m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$

### **Updated Covariance ($P_k$):**
$$
\begin{aligned}
P_k &= P_k^- - K_k \Sigma_{YX} \\
&= P_k^- - K_k H_k P_k^- \\
&= (I - K_k H_k) P_k^-
\end{aligned}
$$

Therefore, the updated conditional distribution simplifies perfectly to:
$$x^{+}_k \sim \mathcal{N}(m_k, P_k)$$

---

## 5. Finding the Conditional Expectation and Variance

By definition, the conditional expectation and conditional variance of the state vector given the observed measurement are the precise mean and covariance parameters of the conditional Gaussian distribution derived in **Question 4**.

### **Conditional Expectation:**
$$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}] = m_k = m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$

### **Conditional Variance:**
$$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) = P_k = (I - K_k H_k) P_k^-$$

where the Kalman Gain $K_k$ remains:
$$K_k = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$

# Scalar Linear-Gaussian Filter Model Derivations

Consider the scalar linear-Gaussian filter model:
$$
\begin{aligned}
x^-_k &= a\,x^+_{k-1} + w_{k-1},\qquad w_{k-1}\sim\mathcal{N}(0,q),\\
y^-_k &= h\,x^-_k + z_k,\qquad\;\;\;\; z_k\sim\mathcal{N}(0,r),
\end{aligned}
$$
with prior $x_0\sim \mathcal{N}(m_0,P_0)$ and $Y_{k}=\{y^{\mathrm{obs}}_1,\dots,y^{\mathrm{obs}}_{k}\}$.

---

## 1. State Prediction Step (Prior Distribution)

We want to derive the mean $m_k^-$ and variance $P_k^-$ of the predicted state $x^-_k$ given the history of observations $Y_{k-1}$.

### Derivation of the Mean $m_k^-$
Taking the conditional expectation of the state update equation with respect to $Y_{k-1}$:
$$
m_k^- = \mathbb{E}[x^-_k \mid Y_{k-1}] = \mathbb{E}[a\,x^+_{k-1} + w_{k-1} \mid Y_{k-1}]
$$

By linearity of expectation and knowing that the process noise $w_{k-1}$ has a zero mean and is independent of past observations $Y_{k-1}$:
$$
m_k^- = a\,\mathbb{E}[x^+_{k-1} \mid Y_{k-1}] + \mathbb{E}[w_{k-1} \mid Y_{k-1}] = a\,m_{k-1} + 0 = a\,m_{k-1}
$$

### Derivation of the Variance $P_k^-$
Taking the conditional variance with respect to $Y_{k-1}$:
$$
P_k^- = \text{Var}(x^-_k \mid Y_{k-1}) = \text{Var}(a\,x^+_{k-1} + w_{k-1} \mid Y_{k-1})
$$

Since the current state $x^+_{k-1}$ and the process noise $w_{k-1}$ are conditionally independent given $Y_{k-1}$, the variance of their sum is the sum of their variances:
$$
P_k^- = \text{Var}(a\,x^+_{k-1} \mid Y_{k-1}) + \text{Var}(w_{k-1} \mid Y_{k-1}) = a^2 P_{k-1} + q
$$

---

## 2. State Measurement Update Step (Posterior Distribution)

The state $x^-_k$ and measurement $y^-_k$ conditioned on $Y_{k-1}$ form a joint Gaussian distribution:
$$
\begin{bmatrix} x^-_k \\ y^-_k \end{bmatrix} \Bigg| Y_{k-1} \sim \mathcal{N} \left( \begin{bmatrix} m_k^- \\ h\,m_k^- \end{bmatrix}, \begin{bmatrix} P_k^- & h\,P_k^- \\ h\,P_k^- & h^2 P_k^- + r \end{bmatrix} \right)
$$

Let the innovation variance be $S_k = h^2 P_k^- + r$. We apply the standard Gaussian conditioning property: if
$$
\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} \mu_X \\ \mu_Y \end{bmatrix}, \begin{bmatrix} \Sigma_{XX} & \Sigma_{XY} \\ \Sigma_{YX} & \Sigma_{YY} \end{bmatrix} \right)
$$
then the conditional distribution $X \mid Y = y^{\mathrm{obs}}$ has mean $\mu_{X|Y} = \mu_X + \Sigma_{XY} \Sigma_{YY}^{-1} (y^{\mathrm{obs}} - \mu_Y)$ and variance $\Sigma_{X|Y} = \Sigma_{XX} - \Sigma_{XY} \Sigma_{YY}^{-1} \Sigma_{YX}$.

Substituting our components ($\mu_X = m_k^-$, $\mu_Y = h\,m_k^-$, $\Sigma_{XX} = P_k^-$, $\Sigma_{XY} = P_k^- h$, and $\Sigma_{YY} = S_k$):

### Kalman Gain ($K_k$)
$$
K_k = \Sigma_{XY} \Sigma_{YY}^{-1} = \frac{P_k^- h}{S_k}
$$

### Posterior Mean ($m_k$)
$$
m_k = m_k^- + K_k (y^{\mathrm{obs}}_k - h\,m_k^-) = m_k^- + \frac{P_k^- h}{S_k}\bigl(y^{\mathrm{obs}}_k - h\,m_k^-\bigr)
$$

### Posterior Variance ($P_k$)
$$
P_k = P_k^- - K_k \Sigma_{YX} = P_k^- - K_k (h\,P_k^-) = (1 - K_k h)P_k^- = \Bigl(1 - \frac{P_k^- h^2}{S_k}\Bigr) P_k^-
$$

---

## 3. Predictive Measurement Distribution

Before processing the actual observation $y^{\mathrm{obs}}_k$, the measurement variable is governed by $y^-_k = h\,x^-_k + z_k$. We evaluate its properties conditioned on $Y_{k-1}$.

### Mean
$$
\mathbb{E}[y^-_k \mid Y_{k-1}] = \mathbb{E}[h\,x^-_k + z_k \mid Y_{k-1}] = h\,\mathbb{E}[x^-_k \mid Y_{k-1}] + \mathbb{E}[z_k] = h\,m_k^- + 0 = h\,m_k^-
$$

### Variance
Since $x^-_k$ and $z_k$ are conditionally independent given $Y_{k-1}$:
$$
\text{Var}(y^-_k \mid Y_{k-1}) = \text{Var}(h\,x^-_k + z_k \mid Y_{k-1}) = h^2 \text{Var}(x^-_k \mid Y_{k-1}) + \text{Var}(z_k) = h^2 P_k^- + r
$$

Thus, the predictive distribution is:
$$
p(y^-_k \mid Y_{k-1})=\mathcal{N}\bigl(h\,m_k^-,\; h^2 P_k^- + r\bigr)
$$

---

## 4. Posterior-Predictive Measurement Distribution

After filtering on the observation $y^{\mathrm{obs}}_k$, our distribution of the true state scales to the posterior $x_k \mid Y_k \sim \mathcal{N}(m_k, P_k)$. A hypothetical new measurement generated at this step follows $y_k = h\,x_k + z_k$.

### Mean
$$
\mathbb{E}[y_k \mid Y_k] = \mathbb{E}[h\,x_k + z_k \mid Y_k] = h\,\mathbb{E}[x_k \mid Y_k] + \mathbb{E}[z_k] = h\,m_k
$$

### Variance
Because measurement noise $z_k$ remains mutually independent of the true state tracking history:
$$
\text{Var}(y_k \mid Y_k) = \text{Var}(h\,x_k + z_k \mid Y_k) = h^2 \text{Var}(x_k \mid Y_k) + \text{Var}(z_k) = h^2 P_k + r
$$

Thus, the posterior-predictive distribution yields:
$$
p(y^-_k\mid Y_k)=\mathcal{N}\bigl(h\,m_k,\; h^2 P_k + r\bigr)
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# 1. Define model parameters
np.random.seed(42)
num_steps = 20
a = 0.95   # State transition coefficient
q = 0.3    # Process noise variance
h = 1.0    # Measurement coefficient
r = 0.4    # Measurement noise variance

# Prior distribution parameters at k=0
m_0 = 0.0
P_0 = 1.5

# 2. Generate true states and noisy observations
true_x = np.zeros(num_steps)
obs_y = np.zeros(num_steps)

x_current = np.random.normal(m_0, np.sqrt(P_0))
for k in range(num_steps):
    x_current = a * x_current + np.random.normal(0, np.sqrt(q))
    true_x[k] = x_current
    obs_y[k] = h * x_current + np.random.normal(0, np.sqrt(r))

# 3. Pre-calculate Kalman Filtering steps
means_prior = []
vars_prior = []
means_post = []
vars_post = []

m_k = m_0
P_k = P_0

for k in range(num_steps):
    # Prediction Step
    m_k_minus = a * m_k
    P_k_minus = (a ** 2) * P_k + q
    means_prior.append(m_k_minus)
    vars_prior.append(P_k_minus)

    # Update Step
    S_k = (h ** 2) * P_k_minus + r
    K_k = (P_k_minus * h) / S_k
    m_k = m_k_minus + K_k * (obs_y[k] - h * m_k_minus)
    P_k = (1 - K_k * h) * P_k_minus
    means_post.append(m_k)
    vars_post.append(P_k)

# 4. Set up plotting framework for the animation
fig, ax = plt.subplots(figsize=(10, 6))
x_axis = np.linspace(-6, 6, 1000)

line_prior, = ax.plot([], [], 'r-', lw=2, label='Prior Density $p(x_k \\mid Y_{k-1})$')
line_post, = ax.plot([], [], 'b-', lw=2, label='Posterior Density $p(x_k \\mid Y_k)$')
vline_true = ax.axvline(0, color='green', linestyle='--', lw=2, label='True State $x_k$')
vline_obs = ax.axvline(0, color='purple', linestyle=':', lw=2, label='Normalized Obs $y^{obs}_k / h$')

ax.set_xlim(-6, 6)
ax.set_ylim(0, 1.5)
ax.set_xlabel('State ($x$)', fontsize=12)
ax.set_ylabel('Probability Density', fontsize=12)
ax.grid(True, alpha=0.3)
ax.legend(loc='upper right')
title = ax.set_title('', fontsize=14)

def init():
    line_prior.set_data([], [])
    line_post.set_data([], [])
    vline_true.set_visible(False)
    vline_obs.set_visible(False)
    return line_prior, line_post, vline_true, vline_obs

def update(frame):
    m_p, v_p = means_prior[frame], vars_prior[frame]
    m_s, v_s = means_post[frame], vars_post[frame]

    y_prior = norm.pdf(x_axis, m_p, np.sqrt(v_p))
    y_post = norm.pdf(x_axis, m_s, np.sqrt(v_s))

    line_prior.set_data(x_axis, y_prior)
    line_post.set_data(x_axis, y_post)

    vline_true.set_xdata([true_x[frame], true_x[frame]])
    vline_obs.set_xdata([obs_y[frame]/h, obs_y[frame]/h])
    vline_true.set_visible(True)
    vline_obs.set_visible(True)

    title.set_text(f'Kalman Filter Evolution — Time Step k = {frame + 1}')
    return line_prior, line_post, vline_true, vline_obs

anim = FuncAnimation(fig, update, frames=num_steps, init_func=init, blit=True, interval=1200)
plt.close()

# Display JavaScript-rendered animation inside Google Colab output cell
HTML(anim.to_jshtml())
```

The system is defined as a **constant-velocity** model with a time step of $\Delta t$. Using basic kinematics:
* $p_x(k) = p_x(k-1) + v_x(k-1)\Delta t$
* $p_y(k) = p_y(k-1) + v_y(k-1)\Delta t$
* $v_x(k) = v_x(k-1)$
* $v_y(k) = v_y(k-1)$

Writing these equations in the matrix form $x_k = A x_{k-1}$:

$$
\begin{bmatrix}
p_x(k)\\
p_y(k)\\
v_x(k)\\
v_y(k)
\end{bmatrix}
=
\begin{bmatrix}
1 & 0 & \Delta t & 0\\
0 & 1 & 0 & \Delta t\\
0 & 0 & 1 & 0\\
0 & 0 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
p_x(k-1)\\
p_y(k-1)\\
v_x(k-1)\\
v_y(k-1)
\end{bmatrix}
$$

Thus:
$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0\\
0 & 1 & 0 & \Delta t\\
0 & 0 & 1 & 0\\
0 & 0 & 0 & 1
\end{bmatrix}
$$

---

### 2. Measurement Matrix ($H$)

The measurement vector $y_k$ consists only of the position components, ignoring the velocities.
* $p_x^{\mathrm{meas}}(k) = 1 \cdot p_x(k) + 0 \cdot p_y(k) + 0 \cdot v_x(k) + 0 \cdot v_y(k)$
* $p_y^{\mathrm{meas}}(k) = 0 \cdot p_x(k) + 1 \cdot p_y(k) + 0 \cdot v_x(k) + 0 \cdot v_y(k)$

Writing this in the matrix form $y_k = H x_k$:

$$
\begin{bmatrix}
p_x^{\mathrm{meas}}(k)\\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}
=
\begin{bmatrix}
1 & 0 & 0 & 0\\
0 & 1 & 0 & 0
\end{bmatrix}
\begin{bmatrix}
p_x(k)\\
p_y(k)\\
v_x(k)\\
v_y(k)
\end{bmatrix}
$$

Encoding the extraction logic yields:
$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0\\
0 & 1 & 0 & 0
\end{bmatrix}
$$

---

### 3. Process Noise Matrix ($G$)

Assuming the process noise $w_{k-1} = [a_x, a_y]^T$ represents random, constant accelerations acting over the interval $\Delta t$, Newtonian mechanics dictates:
* Position shift: $\Delta p = \frac{1}{2} a \Delta t^2$
* Velocity shift: $\Delta v = a \Delta t$

Mapping these disturbances into the state vector gives $G w_{k-1}$:

$$
G w_{k-1} =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\\
0 & \frac{1}{2}\Delta t^2\\
\Delta t & 0\\
0 & \Delta t
\end{bmatrix}
\begin{bmatrix}
a_x\\
a_y
\end{bmatrix}
$$

Thus, the noise coupling matrix is:
$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\\
0 & \frac{1}{2}\Delta t^2\\
\Delta t & 0\\
0 & \Delta t
\end{bmatrix}
$$

## Part B

Develop a python based method to filter a sequence of noisy GPS obtained position meaurements.

```python
import numpy as np

class GPSKalmanFilter:
    def __init__(self, dt=1.0, process_noise=0.1, measurement_noise=3.0, outlier_threshold=None):
        """
        Initialize the 2D Constant Velocity Kalman Filter.

        :param dt: Time step between measurements (seconds)
        :param process_noise: How much the actual velocity fluctuates (system uncertainty)
        :param measurement_noise: Expected standard deviation of GPS error (meters)
        :param outlier_threshold: Standard deviations allowed before rejecting a measurement
        """
        self.dt = dt
        self.outlier_threshold = outlier_threshold
        self.initialized = False

        # State vector [x, y, vx, vy]^T
        self.X = np.zeros((4, 1))

        # State Covariance matrix (Initial high uncertainty)
        self.P = np.eye(4) * 1000.0

        # State Transition matrix (Physics model: pos = pos + vel * dt)
        self.F = np.array([
            [1, 0, dt,  0],
            [0, 1,  0, dt],
            [0, 0,  1,  0],
            [0, 0,  0,  1]
        ])

        # Measurement Matrix (We only observe position x and y)
        self.H = np.array([
            [1, 0, 0, 0],
            [0, 1, 0, 0]
        ])

        # Process Noise Covariance Q
        # Assuming noise enters through acceleration acceleration_noise * dt
        q = process_noise
        self.Q = np.array([
            [q*dt**3/3,         0, q*dt**2/2,         0],
            [        0, q*dt**3/3,         0, q*dt**2/2],
            [q*dt**2/2,         0,   q*dt,            0],
            [        0, q*dt**2/2,         0,      q*dt]
        ])

        # Measurement Noise Covariance R
        self.R = np.eye(2) * (measurement_noise ** 2)

    def process_measurement(self, z_x, z_y):
        """
        Filters a new GPS measurement.

        :param z_x: Measured X coordinate (or Longitude scaled to meters)
        :param z_y: Measured Y coordinate (or Latitude scaled to meters)
        :return: (filtered_x, filtered_y)
        """
        if not self.initialized:
            # Cold start: assume initial position is the first measurement, zero velocity
            self.X = np.array([[z_x], [z_y], [0.0], [0.0]])
            self.initialized = True
            return z_x, z_y

        # --- 1. PREDICT PHASE ---
        self.X = np.dot(self.F, self.X)
        self.P = np.dot(np.dot(self.F, self.P), self.F.T) + self.Q

        # --- 2. UPDATE PHASE ---
        Z = np.array([[z_x], [z_y]])
        Y = Z - np.dot(self.H, self.X)  # Innovation (Measurement residual)
        S = np.dot(np.dot(self.H, self.P), self.H.T) + self.R  # Innovation covariance

        # Optional: Outlier rejection
        if self.outlier_threshold is not None:
            # Mahalanobis distance calculation
            inv_S = np.linalg.inv(S)
            mahalanobis_dist = np.sqrt(np.dot(np.dot(Y.T, inv_S), Y))[0, 0]
            if mahalanobis_dist > self.outlier_threshold:
                # Reject measurement; rely purely on prediction
                return self.X[0, 0], self.X[1, 0]

        # Kalman Gain
        K = np.dot(np.dot(self.P, self.H.T), np.linalg.inv(S))

        # Update State and Covariance
        self.X = self.X + np.dot(K, Y)
        self.P = self.P - np.dot(np.dot(K, self.H), self.P)

        return self.X[0, 0], self.X[1, 0]
```

```python
if __name__ == "__main__":
    # Simulate a vehicle moving in a straight line at constant velocity
    np.random.seed(42)
    timesteps = 50
    true_vx, true_vy = 2.0, 1.5

    true_x = [i * true_vx for i in range(timesteps)]
    true_y = [i * true_vy for i in range(timesteps)]

    # Add random GPS noise (Gaussian) and a couple of nasty spikes (outliers)
    noisy_x = [x + np.random.normal(0, 2.5) for x in true_x]
    noisy_y = [y + np.random.normal(0, 2.5) for y in true_y]

    # Inject deliberate spikes
    noisy_x[15] += 25.0  # Big jump
    noisy_y[35] -= 20.0  # Big jump

    # Instantiate the filter
    # outlier_threshold=3.0 means reject anything outside ~3 standard deviations
    kf = GPSKalmanFilter(dt=1.0, process_noise=0.2, measurement_noise=2.5, outlier_threshold=3.0)

    filtered_points = []
    for nx, ny in zip(noisy_x, noisy_y):
        fx, fy = kf.process_measurement(nx, ny)
        filtered_points.append((fx, fy))

    # Print a quick reality check comparison
    print(f"{'True (X,Y)':<20} | {'Noisy (X,Y)':<20} | {'Filtered (X,Y)':<20}")
    print("-" * 66)
    for i in [10, 15, 20]: # 15 is our outlier index
        print(f"({true_x[i]:.1f}, {true_y[i]:.1f})".ljust(20) + " | " +
              f"({noisy_x[i]:.1f}, {noisy_y[i]:.1f})".ljust(20) + " | " +
              f"({filtered_points[i][0]:.1f}, {filtered_points[i][1]:.1f})")
```
