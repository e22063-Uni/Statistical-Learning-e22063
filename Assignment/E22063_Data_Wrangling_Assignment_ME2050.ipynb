```python
from data_analysis import DataInspector, PlottingMethods
```

```python
# 1. Initialize Engine Structure
engine = DataInspector()
```

```python
# 2. Upload Data
# For automating this testing context, download and run the Titanic training data
import pandas as pd
print("Fetching Titanic test file dynamically...")
titanic_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
engine.df = pd.read_csv(titanic_url, na_values=['?', 'n/a', 'NULL', ' '])
print("Dataset imported directly into analytical loop.\n")
```

```python
# 3. Present Architecture Summaries
engine.display_summary()
```

```python
# 4. Impute Missing Elements
print("\n--- Starting Data Imputation Phase ---")
engine.handle_missing_values('Age', strategy='median')
engine.handle_missing_values('Embarked', strategy='mode')
```

```python
# 5. Outlier Control Phase
print("\n--- Detecting and Flagging Outliers ---")
engine.handle_outliers('Fare', action='flag')
```

```python
# 6. Run Feature Engineering & Scaling Pipeline
print("\n--- Transforming Continuous Features (Normalization) ---")
normalized_continuous = engine.extract_normalized_numeric_data(['Age', 'Fare'], method='standard')

print("\n--- Transforming Categorical Features (Encoding) ---")
encoded_categorical = engine.extract_normalized_categorical_data(['Sex', 'Embarked'], method='onehot')

# Merge features into a single analytical framework
final_features = engine.merge_features(normalized_continuous, encoded_categorical)
print("\nFinal Formed Machine Learning Prepared Matrix Snapshot View:")
display(final_features.head(10))
```

```python
# 7. Run Advanced Visualization Suites
print("\n--- Triggering Chart Generation Engine Loops ---")
# Univariate exploration
engine.plot_univariate_subplots('Age')

# Multi-variable structural relationship tests
engine.plot_relationship('Pclass', 'Age')      # Cat-Num
engine.plot_relationship('Fare', 'Survived')    # Num-Num (Continuous treated classification spectrum mapping)
engine.plot_relationship('Sex', 'Embarked')     # Cat-Cat

# Categorical Frequency Summary Report Tracking
engine.plot_categorical_frequency('Embarked')

# Comprehensive Unified Heatmap Association Framework Engine Trigger
engine.plot_all_associations_heatmap()
```
