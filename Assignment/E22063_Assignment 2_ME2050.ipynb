In a manufacturing process, the proportion of defective items in a large batch produced on a given day is modeled by a continuous random variable $X$. Since $X$ represents a proportion, it takes values in the interval $[0,1]$. Suppose that $X$ has a Beta distribution with parameters $\alpha=2$ and $\beta=18$, so that its density is

f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{1}(1-x)^{17}, & 0<x<1,\\\
0, & \text{otherwise},

Answer the following:

## Sample Answer

## Beta Distribution Analysis: $\text{Beta}(\alpha=2, \beta=18)$

### 1. Probability Density Function (PDF)
The PDF for a Beta distribution with parameters $\alpha$ and $\beta$ is defined as:
$$f(x; \alpha, \beta) = \frac{1}{B(\alpha, \beta)} x^{\alpha-1} (1-x)^{\beta-1}$$

For $\alpha=2$ and $\beta=18$:
$$f(x) = \frac{1}{B(2, 18)} x^{1} (1-x)^{17}, \quad x \in [0, 1]$$

---

### 2. Theoretical Moments
The expected value and variance are calculated as follows:

*   **Expected Value ($E[X]$):**
    $$E[X] = \frac{\alpha}{\alpha + \beta} = \frac{2}{2 + 18} = 0.1$$

*   **Variance ($\text{Var}(X)$):**
    $$\text{Var}(X) = \frac{\alpha\beta}{(\alpha + \beta)^2(\alpha + \beta + 1)} = \frac{(2)(18)}{(20)^2(21)} = \frac{36}{8400} \approx 0.004286$$

---

### 3. Cumulative Probabilities
Probabilities are found using the Regularized Incomplete Beta Function, $I_x(\alpha, \beta)$:

*   **Cumulative Probability $P(X \le 0.10)$:**
    $$P(X \le 0.10) = I_{0.10}(2, 18)$$

*   **Interval Probability $P(0.05 \le X \le 0.15)$:**
    $$P(0.05 \le X \le 0.15) = I_{0.15}(2, 18) - I_{0.05}(2, 18)$$

### The plot of the $\beta$-distribution

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import beta

def plot_beta_distribution(alpha, beta_val):
    """
    Plots the Beta distribution PDF for given alpha and beta parameters using Plotly.
    """
    # Create a range of x values from 0 to 1
    x = np.linspace(0, 1, 1000)

    # Calculate the PDF values for the Beta distribution
    y = beta.pdf(x, alpha, beta_val)

    # Create the Plotly figure
    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=x,
        y=y,
        mode='lines',
        name=f'Beta($\\alpha={alpha}, \\beta={beta_val}$)',
        line=dict(width=2)
    ))

    # Update layout for better readability
    fig.update_layout(
        title=f'Beta Distribution Density Function ($\\alpha={alpha}, \\beta={beta_val}$)',
        xaxis_title='x',
        yaxis_title='PDF(x)',
        template='plotly_white'
    )

    # To display the plot, use fig.show()
    return fig

# Example usage:
fig = plot_beta_distribution(alpha=2, beta_val=18)
fig.show()
```

## Sample Answer

# Triangular Distribution Analysis: $Triangular(4, 10, 6)$

This notebook documents the properties and calculations for a Triangular Distribution where:
*   Minimum ($a$) = $4$
*   Maximum ($b$) = $10$
*   Mode ($c$) = $6$

---

### 1. Support of $X$
The random variable $X$ exists within the interval:
$$a \le X \le b \implies 4 \le X \le 10$$

---

### 2. Probability Density Function (PDF)
The PDF $f(x)$ is defined piecewise:

$$
f(x) =
\begin{cases}
\frac{2(x - a)}{(b - a)(c - a)} & \text{for } a \le x < c \\
\frac{2}{b - a} & \text{for } x = c \\
\frac{2(b - x)}{(b - a)(b - c)} & \text{for } c < x \le b
\end{cases}
$$

Substituting our values:
*   For $4 \le x < 6$: $f(x) = \frac{2(x-4)}{(6)(2)} = \frac{x-4}{6}$
*   For $6 < x \le 10$: $f(x) = \frac{2(10-x)}{(6)(4)} = \frac{10-x}{12}$

---

### 3. Verification of PDF
To verify the PDF, we ensure the area under the curve equals 1:
$$\int_{4}^{10} f(x)dx = \int_{4}^{6} \frac{x-4}{6} dx + \int_{6}^{10} \frac{10-x}{12} dx$$

**Calculations:**
*   $\text{Part 1: } \left[ \frac{(x-4)^2}{12} \right]_{4}^{6} = \frac{4}{12} = 0.333$
*   $\text{Part 2: } \left[ -\frac{(10-x)^2}{24} \right]_{6}^{10} = 0 - (-\frac{16}{24}) = 0.667$
*   **Total:** $0.333 + 0.667 = 1.0$

---

### 4. Expected Value and Variance
**Expected Value $E[X]$:**
$$E[X] = \frac{a + b + c}{3} = \frac{4 + 10 + 6}{3} = 6.667$$

**Variance $Var(X)$:**
$$Var(X) = \frac{a^2 + b^2 + c^2 - ab - ac - bc}{18}$$
$$Var(X) = \frac{16 + 100 + 36 - 40 - 24 - 60}{18} = \frac{28}{18} \approx 1.556$$

---

### 5. Probability Calculations

#### $P(X \le 5)$
Using the first part of the PDF:
$$P(X \le 5) = \int_{4}^{5} \frac{x-4}{6} dx = \left[ \frac{(x-4)^2}{12} \right]_{4}^{5} = \frac{1}{12} \approx 0.0833$$

#### $P(5 \le X \le 8)$
This requires splitting the integral at the mode ($c=6$):
$$P(5 \le X \le 8) = \int_{5}^{6} \frac{x-4}{6} dx + \int_{6}^{8} \frac{10-x}{12} dx$$

1.  **First Integral:** $\left[ \frac{(x-4)^2}{12} \right]_{5}^{6} = \frac{4}{12} - \frac{1}{12} = 0.25$
2.  **Second Integral:** $\left[ -\frac{(10-x)^2}{24} \right]_{6}^{8} = -\frac{4}{24} - (-\frac{16}{24}) = 0.5$
3.  **Total Probability:** $0.25 + 0.5 = 0.75$

---

### 6. Application
In project management (e.g., PERT), these parameters represent:
*   **Optimistic ($a$):** 4 days
*   **Pessimistic ($b$):** 10 days
*   **Most Likely ($c$):** 6 days

### Plot of the triangular distribution

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang

# Parameters given
a = 4
b = 10
c_mode = 6

# Scipy parametrization
# loc is the minimum (a)
# scale is the width (b - a)
# c_scipy is the ratio of the mode relative to the width (c - a) / (b - a)
loc = a
scale = b - a
c_scipy = (c_mode - a) / (b - a)

# Generate data points
x = np.linspace(0, b, 500)
y = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)

# Create the Plotly figure
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode='lines',
    fill='tozeroy',
    name='Triangular Distribution',
    line=dict(color='royalblue', width=2)
))

# Layout updates
fig.update_layout(
    title=f'Triangular Distribution (a={a}, b={b}, c={c_mode})',
    xaxis_title='Time (hours)',
    yaxis_title='Probability Density',
    template='plotly_white'
)

# Display the plot
fig.show()
```

#Q6) Expected Project Completion Time Using PERT

A project manager is estimating the time required to complete a critical software integration task. Let $X$ denote the number of days required to complete the task. Based on expert judgment, the shortest possible time is $a=8$ days, the longest possible time is $b=20$ days, and the most likely completion time is $c=11$ days. Assume that $X$ follows a PERT distribution with parameters $(a,b,c)$.

The PERT distribution is defined as a Beta distribution scaled to the interval $[a,b]$, with internal shape parameters

$$
\alpha = 1 + 4\left( \frac{c-a}{b-a} \right)
$$

and

$$
\beta = 1 + 4\left( \frac{b-c}{b-a} \right).
$$

Its probability density function is

$$
f(x)=\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}}, \qquad a \leq x \leq b,
$$

where

$$
B(\alpha,\beta)=\int_0^1 t^{\alpha-1}(1-t)^{\beta-1},dt.
$$

Answer the following:

1. Compute the values of the shape parameters $\alpha$ and $\beta$.

2. Specify the support of $X$.

3. Write down the probability density function $f_X(x)$ for this model.

4. State the probability space $(\Omega,\mathfrak F,P)$ suitable for this continuous model.

5. Define $X$ as a random variable on this space.

6. Show that the expectation $E[X]$ is

$$
E[X]=\frac{a+4c+b}{6}.
$$

7. Compute the variance $\mathrm{Var}(X)$ using the approximation

$$
\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7}.
$$

8. Compute the approximate standard deviation $\sigma$.

9. Briefly explain why a PERT distribution may be more appropriate than a Triangular distribution in this setting byt comparing the two results.

10. Interpret the practical meaning of the values $a$, $b$, $c$, $E[X]$, and $\sigma$ in the context of project planning.


## Sample Answer

# PERT Distribution Analysis

## 1. Parameters and Shape
Given the optimistic time ($a$), pessimistic time ($b$), and most likely time ($c$):
*   $a = 8$
*   $b = 20$
*   $c = 11$

The shape parameters $\alpha$ and $\beta$ for the underlying Beta distribution are calculated as:
$$\alpha = 1 + 4 \left( \frac{c - a}{b - a} \right) = 1 + 4 \left( \frac{11 - 8}{20 - 8} \right) = 2.0$$
$$\beta = 1 + 4 \left( \frac{b - c}{b - a} \right) = 1 + 4 \left( \frac{20 - 11}{20 - 8} \right) = 4.0$$

---

## 2. Probability Density Function (PDF)
The random variable $X$ exists on the support $x \in [a, b]$. The PDF is defined using the Beta function $B(\alpha, \beta)$:

$$f(x) = \frac{(x - a)^{\alpha-1} (b - x)^{\beta-1}}{B(\alpha, \beta) (b - a)^{\alpha + \beta - 1}}$$

For these parameters:
$$B(2, 4) = 0.05$$
$$f(x) = 0.00000804 \cdot (x - 8)^{1} \cdot (20 - x)^{3}$$

---

## 3. Statistical Measures (PERT)

### Expectation ($E[X]$)
The expected duration is a weighted average that favors the most likely value $c$:
$$E[X] = \frac{a + 4c + b}{6} = \frac{8 + 4(11) + 20}{6} = 12.0$$

### Variance and Standard Deviation
The variance is calculated relative to the distance between the mean and the boundaries:
$$Var(X) = \frac{(E[X] - a)(b - E[X])}{7} = \frac{(12 - 8)(20 - 12)}{7} \approx 4.5714$$
$$\sigma = \sqrt{Var(X)} \approx 2.138$$

*Note: The standard "Rule of Thumb" $(\frac{b-a}{6})$ yields $2.0$.*

---

## 4. Comparison: PERT vs. Triangular
The PERT distribution provides a smoother "bell" shape compared to the linear "spike" of the Triangular distribution.

| Metric | Triangular Distribution | PERT Distribution |
| :--- | :--- | :--- |
| **Mean** | $\frac{a+b+c}{3} = 13.0$ | $\frac{a+4c+b}{6} = 12.0$ |
| **Variance** | $\frac{a^2+b^2+c^2-ab-ac-bc}{18} = 6.5$ | $\frac{(E[X]-a)(b-E[X])}{7} \approx 4.57$ |
| **Std Dev** | $\approx 2.55$ | $\approx 2.14$ |

---

## 5. Interpretation
*   **Optimistic ($a$):** 8
*   **Pessimistic ($b$):** 20
*   **Most Likely ($c$):** 11
*   **Expected Duration:** 12.0
*   **Uncertainty ($\sigma$):** 2.14

### Plot of the triangle and PERT distributions

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang, beta

# Parameters
a = 4
b = 10
c_mode = 6

# --- Triangular Distribution Calculation ---
# Scipy parametrization for Triangular
c_scipy = (c_mode - a) / (b - a)
loc = a
scale = b - a

# --- PERT Distribution Calculation ---
# PERT is a Beta distribution with specific alpha and beta parameters
# alpha = 1 + 4 * (c - a) / (b - a)
# beta = 1 + 4 * (b - c) / (b - a)
alpha_pert = 1 + 4 * (c_mode - a) / (b - a)
beta_pert = 1 + 4 * (b - c_mode) / (b - a)

# Generate data points
x = np.linspace(0, b, 500)
y_triang = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)
y_pert = beta.pdf(x, alpha_pert, beta_pert, loc=a, scale=b-a)

# Create the Plotly figure
fig = go.Figure()

# Add Triangular Distribution
fig.add_trace(go.Scatter(
    x=x,
    y=y_triang,
    mode='lines',
    name='Triangular Distribution',
    line=dict(color='royalblue', width=2, dash='dash')
))

# Add PERT Distribution
fig.add_trace(go.Scatter(
    x=x,
    y=y_pert,
    mode='lines',
    name='PERT Distribution',
    line=dict(color='firebrick', width=2)
))

# Layout updates
fig.update_layout(
    title=f'Comparison: Triangular vs PERT (a={a}, b={b}, c={c_mode})',
    xaxis_title='Time (hours)',
    yaxis_title='Probability Density',
    template='plotly_white'
)

# Display the plot
fig.show()
```

Use the PERT mean formula

E[X]=\frac{a+4c+b}{6}.

In general, if $E[X]=c$ for a PERT distribution, then

$$
\frac{a+4c+b}{6}=c

which implies

$$
a+b=2c

and therefore

$$
b=2c-a.

Thus

$$
b=17.

Using the approximation

$$
\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7},

and imposing $E[X]=c$, we get

$$
\mathrm{Var}(X)=\frac{(c-a)(b-c)}{7}.
$$

Since $b=2c-a$, this becomes

$$
\mathrm{Var}(X)=\frac{(c-a)^2}{7}.
$$

Equivalently,

$$
\mathrm{Var}(X)=\frac{(b-a)^2}{28}.
$$



If the expected value equals the most likely value, then the model becomes symmetric
